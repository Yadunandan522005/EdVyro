```python
# Check missing values in each column

missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
# Calculate percentage of missing values

missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": missing_percentage
})

missing_summary
```

```python
# Check for duplicate rows

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# Display unique values for categorical columns

categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    print("\n", column)
    print(df[column].unique())
```

```python
# Identify numerical columns

numerical_columns = df.select_dtypes(include=np.number).columns

print("Numerical columns:")
print(list(numerical_columns))
```

```python
# Generate summary statistics for numerical variables

df[numerical_columns].describe()
```

```python
# Generate summary statistics for categorical variables

df[categorical_columns].describe()
```

```python
# Check minimum and maximum values

for column in numerical_columns:
    print(
        f"{column}: "
        f"Minimum = {df[column].min()}, "
        f"Maximum = {df[column].max()}"
    )
```

```python
# Check for invalid age values

invalid_age = df[(df["Age"] < 18) | (df["Age"] > 100)]

print("Number of invalid age records:", len(invalid_age))

invalid_age
```

```python
# Check for invalid tenure values

invalid_tenure = df[df["TenureMonths"] < 0]

print("Number of invalid tenure records:", len(invalid_tenure))

invalid_tenure
```

```python
# Check for negative charges

invalid_charges = df[
    (df["MonthlyCharges"] < 0) |
    (df["TotalCharges"] < 0)
]

print("Number of records with negative charges:", len(invalid_charges))

invalid_charges
```

```python
# Check for negative support ticket counts

invalid_tickets = df[df["SupportTickets"] < 0]

print("Number of invalid support-ticket records:", len(invalid_tickets))

invalid_tickets
```

```python
# Detect outliers using the IQR method

outlier_summary = []

for column in numerical_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    outlier_summary.append({
        "Column": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers)
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary
```

```python
# Create boxplots for numerical variables

for column in numerical_columns:
    plt.figure(figsize=(7, 4))

    sns.boxplot(x=df[column])

    plt.title(f"Boxplot of {column}")
    plt.xlabel(column)
    plt.show()
```

```python
# Plot histograms for numerical variables

for column in numerical_columns:
    plt.figure(figsize=(7, 4))

    sns.histplot(df[column], kde=True)

    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.show()
```

```python
# Count churn categories

print(df["Churn"].value_counts())
```

```python
# Calculate churn percentages

churn_percentage = df["Churn"].value_counts(normalize=True) * 100

print(churn_percentage)
```

```python
# Visualize churn distribution

plt.figure(figsize=(6, 4))

sns.countplot(data=df, x="Churn")

plt.title("Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")

plt.show()
```

```python
# Plot categorical variable distributions

categorical_features = [
    "Gender",
    "SubscriptionType",
    "ContractType",
    "PaymentMethod"
]

for column in categorical_features:
    plt.figure(figsize=(7, 4))

    sns.countplot(data=df, x=column)

    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Number of Customers")

    plt.xticks(rotation=30)
    plt.show()
```

```python
# Analyze churn by contract type

contract_churn = pd.crosstab(
    df["ContractType"],
    df["Churn"]
)

contract_churn
```

```python
# Visualize churn by contract type

contract_churn.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Customer Churn by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Number of Customers")

plt.xticks(rotation=0)
plt.legend(title="Churn")

plt.show()
```

```python
# Analyze churn by subscription type

subscription_churn = pd.crosstab(
    df["SubscriptionType"],
    df["Churn"]
)

subscription_churn
```

```python
# Visualize churn by subscription type

subscription_churn.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Customer Churn by Subscription Type")
plt.xlabel("Subscription Type")
plt.ylabel("Number of Customers")

plt.xticks(rotation=0)
plt.legend(title="Churn")

plt.show()
```

```python
# Analyze churn by payment method

payment_churn = pd.crosstab(
    df["PaymentMethod"],
    df["Churn"]
)

payment_churn
```

```python
# Create a copy for cleaning

cleaned_df = df.copy()

# Remove leading/trailing spaces from text columns
for column in cleaned_df.select_dtypes(include="object").columns:
    cleaned_df[column] = cleaned_df[column].str.strip()

print("Cleaning completed.")
```

```python
# Validate missing values after cleaning

print("Missing values after cleaning:")
print(cleaned_df.isnull().sum())

print("\nDuplicate rows after cleaning:")
print(cleaned_df.duplicated().sum())

print("\nDataset shape after cleaning:")
print(cleaned_df.shape)
```

```python
# Final validation checks

print("VALIDATION RESULTS")
print("-" * 40)

print("Rows:", cleaned_df.shape[0])
print("Columns:", cleaned_df.shape[1])
print("Missing values:", cleaned_df.isnull().sum().sum())
print("Duplicate rows:", cleaned_df.duplicated().sum())
print("Duplicate Customer IDs:", cleaned_df["CustomerID"].duplicated().sum())

print("-" * 40)
print("Validation completed successfully.")
```

```python
# Save the cleaned dataset

output_file = "customer_churn_cleaned.csv"

cleaned_df.to_csv(output_file, index=False)

print(f"Cleaned dataset saved as: {output_file}")
```
